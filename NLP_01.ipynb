## 4. Install Gensim and Create a Bag-of-Words Corpus

```python
!pip install gensim
```

```python
from gensim import corpora

# 1. Sample tokenized documents
documents = [
    ["natural", "language", "processing"],
    ["language", "models", "word", "embeddings"]
]

# 2. Build dictionary
dictionary = corpora.Dictionary(documents)

# 3. Convert documents to Bag-of-Words vectors (the corpus)
corpus = [dictionary.doc2bow(doc) for doc in documents]

print(corpus)
```

```python
corpus = [[(0, 1), (1, 1), (2, 1)], [(0, 1), (3, 1), (4, 1), (5, 1)]]

# Print actual words for the first document:
doc_0 = corpus[0]
decoded_doc = [(dictionary[id], count) for id, count in doc_0]

print(decoded_doc)
```
